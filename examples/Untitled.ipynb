```python
def find_in_dict(d, target, path=None, results=None):
    if results is None:
        results = []
    if path is None:
        path = []

    for k, v in d.items():
        # Check the value
        if isinstance(v, str):
            if target in v:
                results.append(path + [k])

        # Recurse if value is another dict
        if isinstance(v, dict):
            find_in_dict(v, target, path + [k], results)

    return results

find_in_dict(ds.attrs, "39WVU")
```
