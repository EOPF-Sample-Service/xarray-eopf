### Open as Dataset

The function `xarray.open_dataset(path, engine="eopf-zarr", op_mode="native", **kwargs)` flattens the `DataTree` structure and returns a single `xr.Dataset`.

In this process, hierarchical groups within the Zarr product are removed by converting their contents into standalone datasets and merging them into one. To ensure uniqueness, variable and dimension names are prefixed with their original group paths, using an underscore (`_`) as the default separator. For example, a variable named `b02` located in the group `measurements/reflectance/r10m` will be renamed to `measurements_reflectance_r10m_b02` in the returned dataset.

```python
%%time
ds = xr.open_dataset(path, engine="eopf-zarr", op_mode="native", chunks={})
ds
```

```python
ds.measurements_r10m_b02
```

```python
%%time
ds.measurements_r60m_b09.plot(vmin=0.0, vmax=0.3)
```

### Open as Dataset in Analysis mode

The **analysis mode** is designed to present EOPF data products in an **analysis-ready and user-friendly** format using xarray’s `Dataset` data model. For Sentinel-2 products, it applies a unified grid mapping across all data variables, meaning selected variables are spatially upscaled or downscaled as needed. This ensures the resulting dataset uses a single, consistent pair of `x` and `y` coordinates for all variables.

```python
%time
ds = xr.open_dataset(path, engine="eopf-zarr", resolution=60, chunks={})
ds
```

```python
%%time
ds.b01.plot.imshow(vmin=0.0, vmax=0.3)
```

```python
%time
ds = xr.open_dataset(
    path,
    engine="eopf-zarr",
    resolution=10,
    bbox=[340000, 4900000, 380000, 4940000],
    chunks={},
)
ds
```

```python
%%time
ds.b01.plot.imshow(vmin=0.0, vmax=0.3)
```

```python
ds = xr.open_dataset(
    path,
    engine="eopf-zarr",
    resolution=0.0002,
    crs="EPSG:4326",
    bbox=[7.2, 44.2, 7.6, 44.6],
    chunks={},
)
ds
```

```python
%%time
ds.b01.plot.imshow(vmin=0.0, vmax=0.3)
```

## Accessing Sub-Groups of a DataTree

This enables referencing **sub-groups within a Zarr-based `DataTree`** directly via STAC assets.

For example, the following STAC item references a `60m` resolution subgroup of a Sentinel-2 L2A dataset:  
👉 [S2B_MSIL2A STAC item with 60m subgroup access](https://stac.browser.user.eopf.eodc.eu/collections/sentinel-2-l2a/items/S2B_MSIL2A_20250430T101559_N0511_R065_T32TLQ_20250430T131328?.language=en&.asset=asset-sr_60m)

In this case, the `asset-sr_60m` asset points directly to the corresponding sub-group (`sr_60m`) within the Zarr archive.

```python
path = (
    "https://objectstore.eodc.eu:2222/e05ab01a9d56408d82ac32d69a5aae2a:202504-s02msil2a/30/"
    "products/cpm_v256/S2B_MSIL2A_20250430T101559_N0511_R065_T32TLQ_20250430T131328.zarr"
    "/measurements/reflectance/r60m"
)
```

```python
%%time
ds = xr.open_dataset(path, engine="eopf-zarr", op_mode="native", chunks={})
ds
```

```python
ds.b02
```

```python
%%time
ds.b02.plot.imshow(vmin=0.0, vmax=0.3)
```
