### Open in native mode

```python
import xarray as xr
path = (
    "https://objects.eodc.eu/e05ab01a9d56408d82ac32d69a5aae2a:202510-s03slsrbt-global/16/products/cpm_v256/"
    "S3B_SL_1_RBT____20251016T072510_20251016T072810_20251016T092049_0179_112_163_2700_ESA_O_NR_004.zarr"
)
dt = xr.open_datatree(path, engine="zarr", chunks={})
dt
```

```python
%%time
dt = xr.open_datatree(path, engine="eopf-zarr", chunks={}, op_mode="native")
dt
```

```python
dt.conditions.meteorology.to_dataset()
```

```python
import matplotlib.pyplot as plt

# Assuming you have:
# lat1, lon1 -> arrays of shape (M, N)
# lat2, lon2 -> arrays of shape (P, Q)

# Flatten them for scatter plotting
x1, y1 = dt.conditions.meteorology.longitude.values[::50, :].ravel(), dt.conditions.meteorology.latitude.values[::50, :].ravel()
x2, y2 = dt.measurements.aoblique.longitude.values[::100, ::100].ravel(), dt.measurements.aoblique.latitude.values[::100, ::100].ravel()
x3, y3 = dt.measurements.anadir.longitude.values[::100, ::100].ravel(), dt.measurements.anadir.latitude.values[::100, ::100].ravel()

plt.figure(figsize=(8, 6))

plt.scatter(x1, y1, s=1, alpha=0.5, label='meteo')
plt.scatter(x2, y2, s=4, alpha=0.5, label='oblique')
plt.scatter(x3, y3, s=1, alpha=0.5, label='nadir')

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Lat/Lon Comparison")
plt.legend()
plt.grid(True)

plt.show()
```
